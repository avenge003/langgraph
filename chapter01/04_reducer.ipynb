```python
def my_reducer(left:list[str], right:list[str]) -> list[str]:
    """
    自定义reducer，将left和right合并
    :param left: 从最开始的位置开始，合并到当前节点的位置的值
    :param right: 当前节点的值
    :return: 合并后的值
    """
    return left + right

# 1. node1 运行之后的值
left = ["start", "node_1 运行完毕"]
right = ["node_2 运行完毕"]

# 2. 合并
merged = my_reducer(left, right)
print(merged)
```
